```python
import pandas as pd, re

df_data = pd.read_csv("../data/dane.csv", dtype=str, keep_default_na=False, na_values=[""])

DATE_RE = re.compile(r"^\s*\d{1,2}[/-]\d{1,2}[/-]\d{2,4}([ T]\d{1,2}:\d{2}(:\d{2})?)?\s*$")

date_cols, empty_cols, mixed = [], [], {}
for c in df_data.columns:
    s = df_data[c].dropna()
    if s.empty:
        empty_cols.append(c)          # brak danych -> nie da się orzec
        continue
    hit = s.str.match(DATE_RE).mean()
    if hit == 1.0:
        date_cols.append(c)
    elif hit > 0:
        mixed[c] = hit                # do ręcznego obejrzenia/
```

```python
size_cols = [
    "z34", "z345", "z35", "z355", "z36", "z365",
    "z37", "z375", "z38", "z385", "z39", "z395",
    "z40", "z405", "z41", "z415", "z42", "z425",
    "z43", "z435", "z44", "z445", "z45", "z46",
    "z47", "z48", "z49", "z50"
]

kolumny_do_pominiecia = set(size_cols + ["etap"] + ['ID'])

duplikaty_id = df[df.duplicated("ID", keep=False)].copy()

def rozniace_sie_kolumny(grupa):
    roznice = []

    for kolumna in df.columns:
        if kolumna in kolumny_do_pominiecia:
            continue

        wartosci = grupa[kolumna].astype("string").fillna("").unique()

        if len(wartosci) > 1:
            roznice.append(kolumna)

    return roznice

raport_roznic = (
    duplikaty_id
    .groupby("ID", group_keys=False)
    .apply(rozniace_sie_kolumny)
    .rename("roznice_poza_etapem_i_rozmiarowka")
)

raport_roznic.head()
```
