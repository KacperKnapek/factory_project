```python
DATE_RE = re.compile(r"^\s*\d{1,2}[/-]\d{1,2}[/-]\d{2,4}([ T]\d{1,2}:\d{2}(:\d{2})?)?\s*$")

date_cols, empty_cols, mixed = [], [], {}
for c in df.columns:
    s = df[c].dropna()
    if s.empty:
        empty_cols.append(c)          # brak danych -> nie da się orzec
        continue
    hit = s.str.match(DATE_RE).mean()
    if hit == 1.0:
        date_cols.append(c)
    elif hit > 0:
        mixed[c] = hit                # do ręcznego obejrzenia/
```

```python
size_cols = [
    "z34", "z345", "z35", "z355", "z36", "z365",
    "z37", "z375", "z38", "z385", "z39", "z395",
    "z40", "z405", "z41", "z415", "z42", "z425",
    "z43", "z435", "z44", "z445", "z45", "z46",
    "z47", "z48", "z49", "z50"
]

cols_to_skip = set(size_cols + ["etap"] + ['ID'])

duplikaty_id = df[df.duplicated("ID", keep=False)].copy()

def rozniace_sie_kolumny(grupa):
    roznice = []

    for col in df.columns:
        if col in cols_to_skip:
            continue

        wartosci = grupa[col].astype("string").fillna("").unique()

        if len(wartosci) > 1:
            roznice.append(col)

    return roznice

raport_roznic = (
    duplikaty_id
    .groupby("ID", group_keys=False)
    .apply(rozniace_sie_kolumny)
    .rename("roznice_poza_etapem_i_rozmiarowka")
)

raport_roznic.head()
```

```python
for c in date_cols:
    df[c] = pd.to_datetime(df[c], errors='coerce', format ='%m/%d/%y %H:%M:%S')
```

```python
date_cols = [c for c in date_cols if c != 'pole1']
if 'przygotowane komplety' not in date_cols:
    date_cols.append('przygotowane komplety')
```

```python
for poprzedni, nastepny in zip(date_cols, date_cols[1:]):
    blad = (
        df[poprzedni].notna()
        & df[nastepny].notna()
        & (df[nastepny] < df[poprzedni])
    )

    if blad.any():
        print(f'Błąd: {nastepny} jest przed {poprzedni}')
        display(df.loc[blad, ['ID', poprzedni, nastepny]])
```
